```python
#klasör oluşturup resimleri yüklebilirsin
import os
os.makedirs("frames",exist_ok=True) #
```

```python
from PIL import Image
frames = []
#resimlerin hangi klasörğn içinde olduğunu bir değişkene atıyoruz.
folder = "frames"
#klasörde sadece resimler olmayabilir.
for file in sorted(os.listdir(folder)):
    if file.endswith(".png") or file.endswith(".jpg"):
        #dosya yolunu oluştur,resmi aç
        img = Image.open(os.path.join(folder, file))
        frames.append(img)
#gif olarak kaydetme
frames[0].save(
    "output.gif",
    save_all=True,
    append_images=frames[1:],
    duration=500,
    loop=0
)
#gif görüntüle
from IPython.display import Image
Image(filename="output.gif")
#indir
from google.colab import files
files.download("output.gif")
```
